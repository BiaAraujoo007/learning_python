```
def calcular_raiz(radicando, radical):
  return radicando ** (1 / radical)
print(calcular_raiz(4, 2))
print(calcular_raiz(8, 3))
print(calcular_raiz(16, 4))
```

```
# 5! = 5x4x3x2x1
def calcular_fatorial(n):
  if n < 2:
    return 1
  result = n
  for i in range(2,n):
    result *= i
  return result

print(calcular_fatorial(5))
```

```
# 5! = 5x4x3x2x1
def calcular_fatorial(n):
  if n < 2:
    return 1
  return n * calcular_fatorial(n-1)

print(calcular_fatorial(5))
```
